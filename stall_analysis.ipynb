**TASK 1**

In this section,we initialize the daily sales dataset to determine the exact name and week for the best and worst sale perfomances

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

    "Onions":  {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

    "Sukuma":  {"qty": 9,"reorder": 15, "cost": 30, "price": 50},

    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},

    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
total_sales = sum(daily_sales)
avg_sales = total_sales / len(daily_sales)
Max_sales = max(daily_sales)
Min_sales = min(daily_sales)
Max_sales_index = daily_sales.index(Max_sales)
Min_sales_index = daily_sales.index(Min_sales)
best_day_name = days[Max_sales_index]
best_week = (Max_sales_index // 7) + 1
worst_day_name = days[Min_sales_index]
worst_week = (Min_sales_index // 7) + 1

print(f"total sales:KES{total_sales}")
print(f"average daily sales:KES{avg_sales:,.2f}")
print(f"best day: {best_day_name} in week {best_week}(KES{Max_sales})")
print(f"worst day: {worst_day_name} in week {worst_week}(KES{Min_sales})")
```

**TASK 2**

We define a reusable function week_totals(sales) that slices the 14-day list into two 7-day blocks, computes their sums, and returns a tuple of both weekly totals. We then unpack these values to calculate the percentage growth rate between Week 1 and Week 2.

```python
def week_totals(sales):
  """returns a tuple of (week 1 total,week 2 total) using list slicing."""
  week1_total =sum(sales[:7])
  week2_total =sum(sales[7:])
  return week1_total,week2_total
week1_total,week2_total = week_totals(daily_sales)
growth_rate = (week2_total - week1_total) / week1_total * 100

print(f"Week 1 total:KES{week1_total:,.2f}")
print(f"Week 2 total:KES{week2_total:,.2f}")
print(f"Growth rate: {growth_rate:,.2f}%")
```

**TASK 3**

 classify_day(amount, average) to categorize each daily sales figure relative to the 14-day average using defined thresholds (+20% for Strong, -20% for Weak, Normal otherwise). We iterate through the sales records and utilize a dictionary counting pattern to tally the frequencies.

```python
def classify_day(amount, average):
    """classifies a sales day as strong, normal, or weak relative to average."""
    if amount > average * 1.2:
        return "strong"
    elif amount < average * 0.8:
        return "weak"
    else:
        return "normal"

categories = {"strong": 0, "normal": 0, "weak": 0}
for sale in daily_sales:
    category = classify_day(sale, avg_sales)
    categories[category] += 1

print("day classifications count:", categories)
```

**TASK 4**

separating sales data into weekday (Mon-Fri) and weekend (Sat-Sun) buckets by evaluating day labels. We calculate average revenue for both periods and determine the percentage premium generated during weekends.

```python
weekday_sales = [daily_sales[i] for i in range(len(daily_sales)) if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]]
weekend_sales = [daily_sales[i] for i in range(len(daily_sales)) if days[i] in ["Sat", "Sun"]]
avg_weekday_sales = sum(weekday_sales) / len(weekday_sales)
avg_weekend_sales = sum(weekend_sales) / len(weekend_sales)
premium_percentage =((avg_weekend_sales - avg_weekday_sales) / avg_weekday_sales *100)

print(f"Average weekday sales:KES{avg_weekday_sales:,.2f}")
print(f"Average weekend sales:KES{avg_weekend_sales:,.2f}")
print(f"Premium percentage: {premium_percentage:,.2f}%")
```

**TASK 5**
Using the stock dictionary, we define a function that iterates over items using .items(), flags products whose quantity is at or below their reorder threshold, and calculates the gross margin percentage (price - cost) / price for each product.


```python
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}

def analyze_stock(stock_dict):
    """Returns products at/below reorder level and computes gross margins percentage."""
    reorder_list = []
    margins = {}
    for product, info in stock_dict.items():
        if info["qty"] <= info["reorder"]:
            reorder_list.append(product)
        margin_pct = ((info["price"] - info["cost"]) / info["price"]) * 100
        margins[product] = margin_pct
    return reorder_list, margins

reorder_items, gross_margins = analyze_stock(stock)

print(f"Items Needing Restock: {reorder_items}")
print("Gross Margins:")
for prod, m in gross_margins.items():
    print(f"  - {prod}: {m:.1f}%")
```

**TASK 6**

We pull all computed metrics together into a clean, professionally aligned text console report using Python f-strings with thousands separators and two-decimal currency formatting for KES.


```python
print("=" * 45)
print("       STALL PERFORMANCE REPORT        ")
print("=" * 45)
print(f"Total 14-Day Sales     : KES {total_sales:,.2f}")
print(f"Average Daily Sales    : KES {avg_sales:,.2f}")
print(f"Best Sales Day         : KES {Max_sales:,.2f} ({best_day_name}, Week {best_week})")
print(f"Worst Sales Day        : KES {Min_sales:,.2f} ({worst_day_name}, Week {worst_week})")
print("-" * 45)
print(f"Week 1 Total Sales     : KES {week1_total:,.2f}")
print(f"Week 2 Total Sales     : KES {week2_total:,.2f}")
print(f"Week-on-Week Growth    : {growth_rate:.2f}%")
print("-" * 45)
print(f"Day Classifications    : Strong: {categories['strong']} | Normal: {categories['normal']} | Weak: {categories['weak']}")
print(f"Weekday Average        : KES {avg_weekday_sales:,.2f}")
print(f"Weekend Average        : KES {avg_weekend_sales:,.2f} ({premium_percentage:.1f}% higher)")
print("-" * 45)
print(f"Items Needing Restock  : {', '.join(reorder_items)}")
print("Gross Margins          :")
for prod, margin in gross_margins.items():
    print(f"  • {prod:10s}         : {margin:.1f}% (Cost: KES {stock[prod]['cost']}, Price: KES {stock[prod]['price']})")
print("=" * 45)
```
